```python
import numpy as np # Импорт библиотеки NumPy с общепринятым сокращением.
```

```python
a = np.array([3,2,1]) # Создание массива из обычных списков;
matrix = np.eye(3,3) # Создание массива-единичной матрицы;
ones = np.ones((2,5)) # Создание массива из единиц;
zeros = np.zeros((6,1))
ran = np.arange(15).reshape((5,3))

print(a, matrix, ones, zeros, ran, sep = "\n\n")
```

```python
print(a.shape, type(a), sep = "\n\n") #  Демонстрация атрибутов
```

```python
# Примеры индексирования
print(a[0]) # нулевой элемент
print(a[-3]) # тот же самый элемент

# Примеры срезов
print(a[0:2]) # вывод первых двух элементов
print(a[::2]) # вывод четных индексов
```

```python
# Примеры векторных операций
c = np.array([0,5,1])
print(a+c) # сложение
print(a-c) # вычитание
print(a*c) # умножение
print(a/c) # деление
```

```python
# Примеры базовых статистик
scores = np.array([4,6,9,3,5,10,35,65,29,59,90,100])

print("Среднее: ", scores.mean())
sc_min = scores.min()
print("Минимальное: ", sc_min)
sc_max = scores.max()
print("Максимальное: ", sc_max)
print("Нормализованное: ", (scores - sc_min)/(sc_max-sc_min))
```
